# Supermarket Sales Analysis



```python
# Import the required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the supermarket sales dataset
df = pd.read_csv("supermarket_sales.csv")

# Display a success message
print("\nDataset loaded successfully!")
```

```python
# Display the number of rows and columns
print("Dataset shape:", df.shape)

# Display the first five rows
df.head()
```

```python
# Display all column names
print("Column Names:")
print(df.columns.tolist())

# Display data types and non-null counts
print("\nDataset Information:")
df.info()
```

```python
# Generate summary statistics for numerical columns
df.describe()
```

```python
# Plot the distribution of sales
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Total', bins=20, kde=True)

plt.title('Distribution of Total Sales')
plt.xlabel('Total Sales Amount')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# Calculate total sales for each product line
product_sales = df.groupby('Product line')['Total'].sum().sort_values(ascending=False)

# Create a bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x=product_sales.index, y=product_sales.values)

plt.title('Total Sales by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python

# Calculate total sales for each branch
branch_sales = df.groupby('Branch')['Total'].sum().sort_values(ascending=False)

# Create a bar chart
plt.figure(figsize=(8, 5))
sns.barplot(x=branch_sales.index, y=branch_sales.values)

plt.title('Total Sales by Branch')
plt.xlabel('Branch')
plt.ylabel('Total Sales')
plt.tight_layout()
plt.show()
```

```python

# Count transactions by payment method
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Payment')

plt.title('Transactions by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Number of Transactions')
plt.tight_layout()
plt.show()
```

```python

# Count transactions by customer type
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Customer type')

plt.title('Transactions by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Number of Transactions')
plt.tight_layout()
plt.show()
```

```python

# Calculate total sales for each hour
hourly_sales = df.groupby('Time')['Total'].sum()

# Create a line chart
plt.figure(figsize=(10, 5))
plt.plot(hourly_sales.index, hourly_sales.values, marker='o')

plt.title('Sales by Time of Day')
plt.xlabel('Time')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Plot the distribution of customer ratings
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Rating', bins=10, kde=True)

plt.title('Distribution of Customer Ratings')
plt.xlabel('Customer Rating')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python

# Select numerical columns
numeric_df = df.select_dtypes(include='number')

# Calculate correlation
correlation = numeric_df.corr()

# Create a heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap of Numerical Variables')
plt.tight_layout()
plt.show()
```

```python

# Compare sales amounts across product lines
plt.figure(figsize=(11, 6))

sns.boxplot(data=df, x='Product line', y='Total')

plt.title('Sales Distribution by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python

# Generate key findings from the dataset

print("KEY FINDINGS - SUPERMARKET SALES")
print("-" * 40)

# Total sales
print("Total Sales:", round(df['Total'].sum(), 2))

# Average transaction amount
print("Average Transaction Amount:", round(df['Total'].mean(), 2))

# Average customer rating
print("Average Customer Rating:", round(df['Rating'].mean(), 2))

# Product line with highest total sales
best_product = df.groupby('Product line')['Total'].sum().idxmax()
print("Highest-Selling Product Line:", best_product)

# Branch with highest total sales
best_branch = df.groupby('Branch')['Total'].sum().idxmax()
print("Branch with Highest Sales:", best_branch)

# Most frequently used payment method
popular_payment = df['Payment'].mode()[0]
print("Most Common Payment Method:", popular_payment)

# Most frequent customer type
popular_customer = df['Customer type'].mode()[0]
print("Most Common Customer Type:", popular_customer)
```
